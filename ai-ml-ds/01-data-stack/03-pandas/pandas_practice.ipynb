```python
df = pd.read_csv('Titanic-Dataset.csv')
print(df.info()) #Importing & printing info of a dataset
```

```python
#CLEANING THE DATASET

df.dropna(inplace = True)
print(df.info()) # Deleting the NULL Values

df.drop_duplicates(inplace = True)
print(df.info())  # Deleting the DUPLICATE Values
```
