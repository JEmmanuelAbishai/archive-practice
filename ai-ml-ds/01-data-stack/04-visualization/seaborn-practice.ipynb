```python
#Imports

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#Processing Data

data = pd.read_csv("Titanic-Dataset.csv")
data = data.dropna()
print(data.head())
print(data.info())
```

```python
#Countplot

sns.countplot(
    x="Survived",
    data=data
)

plt.title("Passenger Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
#Survival count by Gender

sns.countplot(
    x="Sex",
    hue="Survived",
    data=data
)

plt.title("Survival Count by Gender")
plt.xlabel("Gender")
plt.ylabel("Passengers")
plt.legend(["Not Survived", "Survived"])
plt.show()
```

```python
#Histplot

sns.histplot(
    data["Age"],
    bins=20,
    kde=True
)

plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
#Boxplot for Fare vs Passenger

sns.boxplot(
    x="Pclass",
    y="Fare",
    data=data
)

plt.title("Fare Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.show()
```

```python
#Heatmap

numeric_data = data[
    [
        "PassengerId",
        "Survived",
        "Pclass",
        "Age",
        "SibSp",
        "Parch",
        "Fare"
    ]
]

correlation = numeric_data.corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```
