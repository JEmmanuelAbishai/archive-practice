```python
# Basic Imports

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)
```

```python
# Perceptron
def step(x):
  return np.where(x >= 0, 1, 0)

def perceptron(x, w, b):
  z = np.dot(x, w) + b
  return step(z)

```
