```python
#LAMBDA FUNCTIONS

def func(n):
  return lambda a : a * n

double = func(2)
triple = func(3)

print(double(20))
print(triple(22))
```

```python
n = [1,2,3,4,5,6]
doubled = list(map(lambda x:x * 2, n)) # Lambda w/ map()
print(doubled)

n2 = [1,2,3,4,5,6]
odd = list(filter(lambda x:x % 2 != 0, n2)) # Lambda w/ filter()
print(odd)

s = [("Emil", 25), ("Tobias", 22), ("Linus", 28)]
sort_s = sorted(s, key=lambda x: x[1])              # Lambda w/ sort()
print(sort_s)

word = ["apple", "kiwi", "cherry", "orange"]
s_word = sorted(word, key=lambda x: len(x)) # Lambda w/ sort() to
print(s_word)
```
