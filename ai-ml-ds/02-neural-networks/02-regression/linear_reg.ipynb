```python
# Basic Imports

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)
```

```python
# Creating Data

X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 6, 8, 10], dtype=float)
```

```python
# Linear Model

def lin_model(X, w, b):
  return w * X + b

w = 1.5
b = 0.5

pred = lin_model(X, w, b)
print(pred)
```

```python
# Visualization

plt.scatter(X, y, label = "Scatter Data")
plt.plot(X, pred, color ="red", label = "Pred")

plt.legend()
plt.show()
```
