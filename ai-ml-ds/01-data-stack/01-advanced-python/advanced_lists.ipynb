```python
#ADVANCED LISTS

l1 = [10,20,30,40,50]
l1.reverse()
print(l1) #reversed list

def fil(n):
  if n > 20:
    return n

l2 = [2,45,6,7,23,34]
l2f = list(filter(fil,l2)) #Applying Filters on list
print(l2f)

def squaring(n):
  return n**2

l3 = [16,2,30,12]
s_l3 = list(map(squaring,l3)) #Modifying a List
print(s_l3)

c_list = list(zip(l1,l2)) #Combining Lists
print(c_list)

l4 = [1,1,2,3,4,4,4,5,6]
fval = max(set(l4), key = l4.count) #Most frequent value
print(fval)

```
