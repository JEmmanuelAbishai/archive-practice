```python
# Imports

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generating Data

x = np.array([1,2,3,4,5])
y = np.array([2,4,6,8,10])
```

```python
# Creating the model

def lin_model(x,w,b):
  return w * x + b

w = 2
b = 1

pred = lin_model(x,w,b)
print(pred)
```

```python
# Test Prediction vs Actual Prediction

print("Test: ", pred)
print("Actual: ", y)
```

```python
# Calculating the errors

error = y - pred
print("Error: ", error)
```

```python
# Loss Function

sqr_error = (y - pred) ** 2
print("Squared Error: ", sqr_error)

loss = np.mean(sqr_error)
print("Loss: ", loss)
```

```python
# Gradient for Weights

n = len(x)
dw = (-2/n) * np.sum(x *(y-pred))
print("dw: ", dw)
```

```python
# Gradient for Biases

db = (-2/n) * np.sum(y-pred)
print("db: ", db)
```

```python
# New Params

lr = 0.01
w -= lr * dw
b -= lr * db

print("Loss: ", loss)
print("dw: ", dw)
print("db: ", db)
print("new w: ", w)
print("new b: ", b)
```

```python
# Full Gradient Calculation Loop

i = np.array([2,6,4,7,1])
j = np.array([4,8,6,12,2])

w = 0.0
b = 0.0

lr = 0.01
epochs = 1000
losses = []

n = len(i)

for epoch in range(epochs):
  pred = w * i + b
  loss = np.mean((j-pred) ** 2)
  losses.append(loss)

  dw = (-2/n) * np.sum(i * (j-pred))
  db = (-2/n) * np.sum(j-pred)

  w -= lr * dw
  b -= lr * db
```

```python
# Gradient Outputs

print("w: ", w)
print("b: ", b)
print("Final Loss: ", losses[-1])
```

```python
import matplotlib.pyplot as plt

plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss")
plt.show()
```
