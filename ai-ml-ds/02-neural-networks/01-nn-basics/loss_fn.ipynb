```python
# Imports

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([1,2,3,4,5] , dtype = float)
y = np.array([6,7,8,9,10], dtype = float)

def lin_model(x,w,b):
  return w * x + b
```

```python
# Start Predictions

w = 1.5
b = 0.5

pred = lin_model(x,w,b)
print(pred)
```

```python
# Finding Individual Loss

def ind_loss(y,pred):
  return (y-pred) ** 2

indi_loss = ind_loss(y,pred)
print(indi_loss)
```

```python
# Loss Function (MSE)

def mse(y,pred):
  return np.mean(y - pred) ** 2

loss = mse(y,pred)
print(loss)
```
