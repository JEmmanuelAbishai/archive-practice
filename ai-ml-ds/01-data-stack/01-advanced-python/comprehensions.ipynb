```python
#COMPREHENSIONS

# List comprehensions

a = [1, 2, 3, 4, 5, 6, 7, 8, 9]
res = [num for num in a if num % 2 == 0]
res2 = [num**2 for num in range(1, 6)]
print(res)
print(res2)
```

```python
# Dictionary comprehensions

a = ["Corvette", "McLaren", "Aston Martin"] # brand
b = ["C6", "P1", "DB9"] # car

res = {brand: car for brand, car in zip(a, b)}
print(res)
```

```python
#Set comprehensions

a = [1, 2, 2, 3, 4, 4, 5, 6, 6, 7]
res = {num for num in a if num % 2 == 0}
print(res)

res = {num**2 for num in range(1, 6)}
print(res)
```

```python
#Generator comprehensions

res = (num for num in range(10) if num % 2 == 0)
print(list(res))
```
